## 2. Load and Preprocess Data

We'll load sample invoice data and prepare it for analysis.

```python
# Sample data structure (we'll replace this with real data)
sample_data = {
    'invoice_id': range(1000, 1010),
    'vendor_id': ['V' + str(i) for i in range(101, 111)],
    'amount': np.random.uniform(1000, 10000, 10),
    'date': pd.date_range(start='2025-01-01', periods=10),
    'payment_terms': np.random.choice(['NET30', 'NET45', 'NET60'], 10),
    'status': np.random.choice(['Pending', 'Approved', 'Paid'], 10)
}

df = pd.DataFrame(sample_data)
df.head()
```

## 3. Implement Fraud Detection Rules

Define and apply rules to identify potential fraud indicators:

```python
def check_duplicate_invoices(df):
    """Check for duplicate invoice numbers from the same vendor"""
    duplicates = df[df.duplicated(['vendor_id', 'invoice_id'], keep=False)]
    return duplicates

def check_unusual_amounts(df):
    """Identify unusually high invoice amounts"""
    mean = df['amount'].mean()
    std = df['amount'].std()
    threshold = mean + (2 * std)
    return df[df['amount'] > threshold]

# Apply rules
duplicate_invoices = check_duplicate_invoices(df)
unusual_amounts = check_unusual_amounts(df)

print("Potential fraudulent activities found:")
print(f"Duplicate invoices: {len(duplicate_invoices)}")
print(f"Unusual amounts: {len(unusual_amounts)}")
```
